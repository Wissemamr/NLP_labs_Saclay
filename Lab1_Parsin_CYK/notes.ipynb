`POS : ADJ, ADP, AUX, DET, NOUN, NUM, PROPN, PUNCT, SCONJ, VERB`
- ADJ : adjective 
- ADP : adposition (from, among...),  spatial, temporal, or semantic relations
- AUX : auxiliary : : modal vers conjugates, 
- DET : determinant
- NOUN : noun
- NUM : numercial values, even if not instanticated like n or k counters
- PROPN : proper noun : nom propre like names of people and places
- PUNCT : puntucation
- SCONJ : subordinating conjunction (that, if, while..)
- VERB : at anny tense, conjugated,..

```python
import spacy
from IPython.display import display
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

sentences = [
    "Arabidopsis thaliana seedlings exhibit longer hypocotyls when they are grown under high ambient temperature, which is defined as thermomorphogenesis.",
    "A spectrogram of PSN J10354824+3900279 obtained on Dec. 19.33 UT suggests that this is a type-Ia at redshift z 0.044.",
]

for sentence in sentences:
    doc = nlp(sentence)

    print(f"\nSentence: {sentence}")
    print("Entities:")
    for entity in doc.ents:
        print(f"  {entity.text!r:45} {entity.label_}")

    print("\nTokens:")
    print(f"{'Token':30} {'Lemma':25} {'POS':8} {'Dependency':15}")
    print("-" * 82)
    for token in doc:
        print(f"{token.text:30} {token.lemma_:25} {token.pos_:8} {token.dep_:15}")

    display(displacy.render(doc, style="ent", jupyter=True))
```

```python
import spacy
import scispacy
from IPython.display import display
from pathlib import Path
from shutil import copytree
from tempfile import TemporaryDirectory
from spacy import displacy
import en_core_sci_sm

# Load the SciSpaCy model. Its config was written for spaCy 3.7,
# so patch the two old string booleans for spaCy 3.8 in a temporary copy.
# --> We solve the interior problem in spacy implementation by changing the config file of the library
installed_model_path = Path(en_core_sci_sm.__file__).parent / "en_core_sci_sm-0.5.4"
model_temp_dir = TemporaryDirectory()
model_path = Path(model_temp_dir.name) / installed_model_path.name
copytree(installed_model_path, model_path)
config_path = model_path / "config.cfg"
config = config_path.read_text().replace(
    'include_static_vectors = "False"',
    "include_static_vectors = false",
)
config_path.write_text(config)
nlp = spacy.load(model_path)

sentences = [
    "Arabidopsis thaliana seedlings exhibit longer hypocotyls when they are grown under high ambient temperature, which is defined as thermomorphogenesis.",
    "A spectrogram of PSN J10354824+3900279 obtained on Dec. 19.33 UT suggests that this is a type-Ia at redshift z 0.044.",
]

for sentence in sentences:
    doc = nlp(sentence)

    print(f"\nSentence: {sentence}")
    print("Entities:")
    for entity in doc.ents:
        print(f"  {entity.text!r:45} {entity.label_}")

    print("\nTokens:")
    print(f"{'Token':30} {'Lemma':25} {'POS':8} {'Dependency':15}")
    print("-" * 82)
    for token in doc:
        print(f"{token.text:30} {token.lemma_:25} {token.pos_:8} {token.dep_:15}")

    display(displacy.render(doc, style="ent", jupyter=True))
```
